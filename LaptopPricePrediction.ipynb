```python
import numpy as np
import pandas as pd

df = pd.read_csv("/content/laptops.csv")
df.head()
```

```python
df.info()
```

```python
df.shape
```

```python
df.describe()
```

```python
#removing unnessry cloumn
df.drop(columns=['Unnamed: 0','img_link','name'], inplace=True)
df.head()
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
#filling missing values
df['rating']= df['rating'].fillna(df['rating'].mean())
df['no_of_ratings']= df['no_of_ratings'].fillna(df['no_of_ratings'].mean())
df['no_of_reviews']= df['no_of_reviews'].fillna(df['no_of_reviews'].mean())

df.isna().sum()
```

```python
#feature engineering
#ram conversion
df['ram'] = df['ram'].astype(str).str.extract(r'(\d+)').astype(int)
```

```python
#storage conversion
df['storage'] = df['storage'].astype(str)
df['storage'] = df['storage'].replace("TB","000")
df['storage'] = df['storage'].astype(str).str.extract(r'(\d+)').astype(int)
```

```python
#processor break down
df['processor_brand'] = df['processor'].apply(lambda x: x.split()[0])#splits the input
df['processor_type'] = df['processor'].str.extract(r'(i\d)')
df['generation'] = df['processor'].str.extract(r'(\d+)').astype(float)
```

```python
df['generation'] = df['generation'].fillna(df['generation'].median())
```

```python
#storage type
df['is_ssd'] = df['storage'].apply(lambda x: 1 if 'SSD' in str(x) else 0)# if 1 ssd if 0 then hhd
df['is_hdd'] = df['storage'].apply(lambda x: 1 if 'HDD' in str(x) else 0)#if 1 hhd if 0 then ssd
```

```python
#Drop original column
df.drop(columns=['processor'], inplace=True)
```

```python
#conversion of categorical to numerical
df = pd.get_dummies(df, columns=['processor_brand','processor_type','os'])
```

```python
#defining target variable
x = df.drop('price(in Rs.)',axis =1)
y = df['price(in Rs.)']
```

```python
#Train-Test Split
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
#Model Training
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor

lr = LinearRegression()
rf = RandomForestRegressor(n_estimators=200)

lr.fit(x_train,y_train)
rf.fit(x_train,y_train)
```

```python
#model evaluation
from sklearn.metrics import r2_score

print("linear regression", r2_score(y_test,lr.predict(x_test)))
print("random forest", r2_score(y_test,rf.predict(x_test)))
```

```python
#Hyperparameter Tuning

from sklearn.model_selection import GridSearchCV

params = {
    'n_estimators': [100,200],
    'max_depth' : [5,10,None],
}

grid = GridSearchCV(RandomForestRegressor(), params,cv=3)#cv - cross validation
grid.fit(x_train,y_train)

#getting the best model
best_model = grid.best_estimator_
```

```python
#save model
import pickle

pickle.dump(best_model,open('model.pkl','wb'))
```

```python
import gradio as gr
import pickle
import numpy as np
import pandas as pd

# Load the trained model
model = pickle.load(open('model.pkl', 'rb'))

# Retrieve the column names from the 'x' DataFrame available in the kernel
# This is crucial to ensure the input to the model has the correct features and order.
# In a production environment, you would typically save these column names alongside your model.
model_columns = globals()['x'].columns.tolist()

def predict_laptop_price(
    ram,
    storage,
    display_inch,
    rating,
    no_of_ratings,
    no_of_reviews,
    generation,
    is_ssd_input,
    processor_brand_input,
    processor_type_input,
    os_input
):
    # Create a dictionary to hold all feature values, initialized to 0
    features_dict = {col: 0 for col in model_columns}

    # Assign numerical inputs
    features_dict['ram'] = ram
    features_dict['storage'] = storage
    features_dict['display(in inch)'] = display_inch
    features_dict['rating'] = rating
    features_dict['no_of_ratings'] = no_of_ratings
    features_dict['no_of_reviews'] = no_of_reviews
    features_dict['generation'] = generation

    # Assign derived boolean features
    features_dict['is_ssd'] = 1 if is_ssd_input == 'SSD' else 0
    features_dict['is_hdd'] = 1 if is_ssd_input == 'HDD' else 0 # If not SSD, assume HDD

    # Assign one-hot encoded categorical features
    # Processor Brand
    processor_brand_col = f'processor_brand_{processor_brand_input}'
    if processor_brand_col in features_dict: # Check if this specific brand was in training data
        features_dict[processor_brand_col] = 1
    else:
        # Handle cases where user selects a brand not seen during training, or map to a default
        # For simplicity, we'll ignore if not found. A more robust app might have an 'Other' category.
        pass

    # Processor Type
    processor_type_col = f'processor_type_{processor_type_input}'
    if processor_type_col in features_dict: # Check if type was in training data
        features_dict[processor_type_col] = 1

    # OS
    os_col = f'os_{os_input}'
    if os_col in features_dict: # Check if OS was in training data
        features_dict[os_col] = 1

    # Convert the dictionary to a DataFrame, ensuring column order
    input_df = pd.DataFrame([features_dict])

    # Ensure the order of columns in input_df matches the model_columns
    # This is critical for correct prediction, especially with one-hot encoding.
    input_df = input_df[model_columns]

    # Make prediction
    prediction = model.predict(input_df)[0]
    return f"₹ {int(prediction):,}"

# Define Gradio Interface components
processor_brand_options = ['Intel', 'AMD', 'M1', 'Qualcomm', 'MediaTek'] # Based on unique values in your preprocessed data
processor_type_options = ['i3', 'i5', 'i7', 'i9', 'Other']
os_options = [
    'Windows 11 Operating System', '64 bit Windows 11 Operating System', 'Windows 10 Operating System',
    '64 bit Windows 10 Operating System', 'DOS Operating System', 'Mac Operating System',
    'Chrome Operating System', 'PrimeOS Operating System', 'Windows 8 Operating System', 'Other'
] # Based on unique values in your preprocessed data


iface = gr.Interface(
    fn=predict_laptop_price,
    inputs=[
        gr.Slider(minimum=4, maximum=64, step=4, value=8, label="RAM (GB)"),
        gr.Slider(minimum=128, maximum=2000, step=128, value=512, label="Storage (GB)"),
        gr.Slider(minimum=10.0, maximum=18.0, step=0.1, value=15.6, label="Display Size (inches)"),
        gr.Slider(minimum=1.0, maximum=5.0, step=0.1, value=4.0, label="Rating"),
        gr.Number(value=718, label="Number of Ratings (avg. was 718)"),
        gr.Number(value=84, label="Number of Reviews (avg. was 84)"),
        gr.Slider(minimum=1, maximum=13, step=1, value=10, label="Processor Generation"),
        gr.Radio(['SSD', 'HDD'], value='SSD', label="Storage Type"),
        gr.Dropdown(processor_brand_options, value='Intel', label="Processor Brand"),
        gr.Dropdown(processor_type_options, value='i5', label="Processor Type"),
        gr.Dropdown(os_options, value='Windows 11 Operating System', label="Operating System")
    ],
    outputs=gr.Textbox(label="Predicted Price"),
    title="Laptop Price Predictor",
    description="Enter laptop specifications to get a predicted price."
)

iface.launch(share=True)
```
